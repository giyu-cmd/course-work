## *XOR*

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import Perceptron
from sklearn.neural_network import MLPClassifier
```

```python
# XOR Dataset
X = np.array([[0,0], [0,1], [1,0], [1,1]])
y = np.array([0,1,1,0])
```

```python
plt.figure(figsize=(10,5))

plt.scatter(X[y==0,0] , X[y==0,1] , marker="o" , color="blue" , s=100 , label="Class 0")
plt.scatter(X[y==1,0] , X[y==1,1] , marker="x" , color="red" , s=100 , label="Class 1")

plt.title("Dataset(XOR)")
plt.xlabel("Feature 01")
plt.ylabel("Feature 02")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Single layer perceptron
p = Perceptron(max_iter=1000, random_state=1) # (epochs , random state) 
p.fit(X,y)

print("Perceptron Prediction : ", p.predict(X))
```

```python
xx, yy = np.meshgrid(np.linspace(-0.5,1.5,200) , np.linspace(-0.5,1.5,200))

Z = p.decision_function(np.c_[xx.ravel() , yy.ravel()]).reshape(xx.shape)
plt.contour(xx, yy, Z, levels=[0] , colors ="purple")

plt.scatter(X[y==0,0] , X[y==0,1] , marker="o" , color="blue" , s=100 , label="Class 0")
plt.scatter(X[y==1,0] , X[y==1,1] , marker="x" , color="red" , s=100 , label="Class 1")

plt.title("Decision boundary")
plt.xlabel("Feature 01")
plt.ylabel("Feature 02")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Multi layer perceptron
mlp = MLPClassifier(
    hidden_layer_sizes=(4,),
    activation='tanh',
    max_iter=2000,
    random_state=1
)

mlp.fit(X,y)

# Print
print("MLP Prediction : ", mlp.predict(X))
```

```python
# Grid of points covering the plot area
xx, yy = np.meshgrid(np.linspace(-0.5, 1.5, 200), np.linspace(-0.5, 1.5, 200))
Z = mlp.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

# Decision boundary (purple line where prediction flips from 0 to 1)
plt.contour(xx, yy, Z, levels=[0.5], colors="purple")

# XOR points: class 0 = "o", class 1 = "x"
plt.scatter(X[y==0, 0], X[y==0, 1], marker="o", color="blue", s=100, label="Class 0")
plt.scatter(X[y==1, 0], X[y==1, 1], marker="x", color="red", s=100, label="Class 1")

plt.legend()
plt.grid(True)
plt.title("XOR with MLP Decision Boundary")
plt.show()
```

## *AND*

```python
# AND Dataset
X_and = np.array([[0,0], [0,1], [1,0], [1,1]])
y_and = np.array([0,0,0,1])
```

```python
# Single layer perceptron
p_and = Perceptron(max_iter=1000, random_state=1)
p_and.fit(X,y)

print("Perceptron Prediction : ", p_and.predict(X_and))
```

```python
# Grid of points covering the plot area
xx, yy = np.meshgrid(np.linspace(-0.5, 1.5, 200), np.linspace(-0.5, 1.5, 200))

# Perceptron score for every grid point (the line is where the score = 0)
Z = p_and.decision_function(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
plt.contour(xx, yy, Z, levels=[0], colors="purple")

# XOR points: class 0 = "o", class 1 = "x"
plt.scatter(X_and[y_and==0, 0], X_and[y_and==0, 1], marker="o", color="blue", s=100, label="Class 0")
plt.scatter(X_and[y_and==1, 0], X_and[y_and==1, 1], marker="x", color="red", s=100, label="Class 1")

plt.grid(True)
plt.legend()
plt.title("AND with Perceptron Decision Boundary")
plt.show()

print(p_and.coef_, p_and.intercept_)
```

```python
# Multi layer perceptron
mlp_and = MLPClassifier(
    hidden_layer_sizes=(4,),
    activation='tanh',
    max_iter=1000,
    random_state=0
)

mlp_and.fit(X_and,y_and)

# Print
print("MLP Prediction : ", mlp_and.predict(X_and))
```

```python
# Grid of points covering the plot area
xx, yy = np.meshgrid(np.linspace(-0.5, 1.5, 200), np.linspace(-0.5, 1.5, 200))
Z = mlp_and.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

# Decision boundary (purple line where prediction flips from 0 to 1)
plt.contour(xx, yy, Z, levels=[0.5], colors="purple")

# XOR points: class 0 = "o", class 1 = "x"
plt.scatter(X_and[y_and==0, 0], X_and[y_and==0, 1], marker="o", color="blue", s=100, label="Class 0")
plt.scatter(X_and[y_and==1, 0], X_and[y_and==1, 1], marker="x", color="red", s=100, label="Class 1")

plt.legend()
plt.grid(True)
plt.title("AND with MLP Decision Boundary")
plt.show()
```
